```python
from PIL import Image

MEM_FILE = "rome_rgb.mem"
IMG_WIDTH = 320
IMG_HEIGHT = 240
OUT_PNG = " Cartonkgp_output.png"

pixels = []

with open(MEM_FILE, "r") as f:
    for line in f:
        line = line.strip()
        if len(line) != 6:
            continue  # skip invalid lines

        r = int(line[0:2], 16)
        g = int(line[2:4], 16)
        b = int(line[4:6], 16)

        pixels.append((r, g, b))

expected_pixels = IMG_WIDTH * IMG_HEIGHT
if len(pixels) != expected_pixels:
    print(f"WARNING: Expected {expected_pixels} pixels, got {len(pixels)}")

img = Image.new("RGB", (IMG_WIDTH, IMG_HEIGHT))
img.putdata(pixels)
img.save(OUT_PNG)
print(f"Image saved as {OUT_PNG}")
```

```python
#!/usr/bin/env python3
from PIL import Image

# ─── EDIT THESE ───────────────────────────────
INPUT_MEM  = "iitkgp.mem"                 # input .mem file
OUTPUT_PNG = "rome.png"          # output image file
WIDTH      = 512                    # image width  in pixels
HEIGHT     = 512                    # image height in pixels
# ──────────────────────────────────────────────

pixels = []

with open(INPUT_MEM, "r") as f:
    for lineno, raw in enumerate(f, 1):
        line = raw.strip()
        if not line or line.startswith("//") or line.startswith("#"):
            continue
        if len(line) != 6:
            print(f"  Warning: skipping line {lineno}: {line!r}")
            continue
        try:
            r = int(line[0:2], 16)
            g = int(line[2:4], 16)
            b = int(line[4:6], 16)
            pixels.append((r, g, b))
        except ValueError:
            print(f"  Warning: bad hex on line {lineno}: {line!r}")

expected = WIDTH * HEIGHT
got      = len(pixels)

if got < expected:
    print(f"  Warning: got {got} pixels, expected {expected} — padding with black.")
    pixels += [(0, 0, 0)] * (expected - got)
elif got > expected:
    print(f"  Warning: got {got} pixels, expected {expected} — truncating.")
    pixels = pixels[:expected]

img = Image.new("RGB", (WIDTH, HEIGHT))
img.putdata(pixels)
img.save(OUTPUT_PNG)

print(f"Done: {INPUT_MEM} -> {OUTPUT_PNG}  ({WIDTH}x{HEIGHT})")
```
